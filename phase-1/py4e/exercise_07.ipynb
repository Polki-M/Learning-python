```python
#Prompt user for integer nubers until it enters "done".
#Once "done" is entered, print the largest number.
#If user enters other than a valid number, catch it with a try.
```

```python
largest = None
smallest = None

while True:
    num = input ("Enter a number: ")
    if num == "done":
        break
    try:
        num = int(num)
    except:
        print ("Invalid input")
        continue

    if largest is None or num > largest:
        largest = num
    if smallest is None or num < smallest:
        smallest = num

print ("Maximum: ", largest)
print ("Minimum: ", smallest)
```
