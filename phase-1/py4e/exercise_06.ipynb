```python
#Prompt user for hours and rate per hour using INPUT to compute gross pay
#Normal rate for h<=40.
# 1.5x hourly rate for h>40
```

```python
def computepay (h, r):
    if h < 0:
        print ("ERROR")
    elif h <= 40:
        return h * r
    else:
        return (40 * r) + (h - 40) * (1.5 * r)

shours = input ("Enter hours: ")
srate = input ("Enter rate: ")
h = float(shours)
r = float(srate)

pay = computepay (h, r)
print ("Pay: ", pay)
```
