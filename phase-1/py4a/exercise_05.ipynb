```python
sscore = input ("Enter score: ")

try:
    score = float(sscore)
except:
    print ("Invalid score")
    quit ()

if score < 0.0 > 1.0:
    print ("ERROR")
    quit ()

if score >= 0.9: print ("Grade: A")
elif score >= 0.8: print ("Grade: B")
elif score >= 0.7: print ("Grade: C")
elif score >= 0.6: print ("Grade: D")
elif score < 0.6: print ("Grade: F")
```
